Pour appeler la fonction, il suffit d'indiquer son nom entre guillemets avec les paramètres que vous souhaitez utiliser entre parenthèses.

```python
def nom_de_la_fonction(personne,arg2):
    pass
```

## 2.2 - Return

La fonction return dans une fonction sert à renvoyer un résultat que vous souhaitez récupérer à la fin d'une fonction. Ici, par exemple, nous récupérons le résultat dans une variable a après l'exécution de la fonction.

```python
def ajout_nombre(num1,num2):
    return num1+num2
```

```python
a=ajout_nombre(2,4)
```

## 2.3 - Break

Dans une boucle, « break » est utilisé pour arrêter une boucle. Ici, nous voulons que la boucle s'arrête lorsque i atteint 10.

```python
def jusque_dix(num):
    for i in range(1,num):
        if i%10==0:
            print("Voici 10 !")
            break
        else:
            print(i)
```

```python
jusque_dix(20)
```

```python
jusque_dix(2)
```

## 2.4 Continue et pass

« continue » passe à la boucle suivante sans terminer ces instructions. Si i est un nombre pair, le programme passe à la boucle suivante sans afficher quoi que ce soit.

```python

for i in range(10):
    if i%2==0:
        continue
    print(i)
```


« pass » ne fait rien, le programme continue d'exécuter toutes les
lignes après le bloc if.


```python

for i in range(10):
    if i%2==0:
        pass
    print(i)
```
