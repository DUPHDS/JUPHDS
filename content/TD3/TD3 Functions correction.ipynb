# Exercice 1

Write a function maximum() that takes a list of numbers as an argument and returns the
maximum of this list of numbers.

```python
def maximum(myList):
    number_max=myList[0]
    for number in myList[1:]:
        if number>number_max:
            number_max=number
    return number_max
```

```python
theList=[1,6,9,11,4]
maximum(theList)
```

# Exercice 3

Write a function that return mentions to students, if the grade is :
 - Less than 10: Bad
 - Between 10 and 12: Fair
 - Between 12 and 14: Good
 - Between 14 and 16: Very good
 - Over 16: Congratulations.

1. First, create the mention() function, which takes a note as argument and returns the associated mention.


```python
def mention(note):
    if note<10:
        return "Bad"
    elif note<12:
        return "Fair"
    elif note<14:
        return "Good"
    elif note<16:
        return "Very good"
    else:
        return "Congratulations"
```

```python
mentionToGrade={"Bad":[0,10],"Fair":[10,12],"Good":[12,14],"Very good":[14,16],"Congratulations":[16,21]}

def gradeToMention(grade):
  for mention in mentionToGrade:
    if grade>=mentionToGrade[mention][0] and grade<mentionToGrade[mention][1]:
      return mention
```

2. Finally, create a function that takes a dictionary of students' names with their associated grades and returns a dictionary of students' names with their associated mentions.


```python
def mentionForStudent(myDict):
    mentionsStudent={}
    for studentName in myDict:
        mentionsStudent[studentName]=mention(myDict[studentName])
    return mentionsStudent
```

1. Write a caculate() function which, when run, asks the user to enter two numbers and asks him to choose between adding or subtracting these two numbers. The function will only stop if the user wants to stop, as it can calculate several times from two numbers. And each calculation result will be stored in a list returned at the end.


```python
def calculate():
    x=0
    results=[]
    while x!="stop":
        x=input("Enter \"stop\" if you want to stop the program.")
        if x!="stop":
            n1=int(input("Enter an integer : "))
            n2=int(input("Enter a second one : "))
            calc=input("Addition [add] ? \nSubstraction [sub] ?")
            if calc=="add":
                results.append(n1+n2)
            elif calc=="sub":
                results.append(n1-n2)
            else:
                print("This is not an option.")
    return results
```

1. Write a function called phone() which, when run, asks the user to fill the phone directory with the names and numbers of people he or she knows. He can add as many as he likes.
2. At the end, the function returns the number of people added.


```python
def phone():
    phoneBook={}
    x=0
    while x!="stop":
        name=input("What is the name of the person ?")
        tel=input("What is her phone number ?")
        phoneBook[name]=tel
        x=input("Do you want to stop ? ")
    return len(phoneBook)
```
