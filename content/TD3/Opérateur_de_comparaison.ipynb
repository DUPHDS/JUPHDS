# Opérateur de comparaison

Il existe plusieurs opérateurs de comparaison.

```python
3>2
```

```python
3<2
```

Vous pouvez utiliser « in » pour vérifier si un élément est présent dans un élément contenant plusieurs éléments.

```python
a=[1,2,3]
```

```python
3 in a
```

```python
4 in a
```

```python
4 not in a
```

L'opérateur « and » renvoie la valeur True si les deux conditions sont remplies, sinon il renvoie la valeur False.

```python
3>2 and 4>2
```

```python
3>2 and 4<2
```

L'opérateur « or » renvoie la valeur True si au moins une des conditions est remplie, sinon il renvoie la valeur False.

```python
3>2 or 4<2
```
