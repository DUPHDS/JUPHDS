## 2.3 - Break

In a loop, “break” is used to stop a loop. Here, we want the loop to stop when i reaches 10.

```python
def until_ten(num):
    for i in range(1,num):
        if i%10==0:
            print("It's 10 !")
            break
        else:
            print(i)
```

```python
until_ten(20)
```

```python
until_ten(2)
```
