# Exercice 1

Écrivez une fonction maximum() qui prend une liste de nombres comme argument et renvoie le
maximum de cette liste de nombres.

```python
def maximum(myList):
    number_max=myList[0]
    for number in myList[1:]:
        if number>number_max:
            number_max=number
    return number_max
```

```python
theList=[1,6,9,11,4]
maximum(theList)
```

1. Commencez par créer la fonction mention(), qui prend une note comme argument et renvoie la mention associée.


```python
def mention(note):
    if note<10:
        return "Mauvaise"
    elif note<12:
        return "Passable"
    elif note<14:
        return "Bonne"
    elif note<16:
        return "Très bonne"
    else:
        return "Félicitations"
```

# Exercice 4

Écrivez une fonction appelée justOne() qui supprime les doublons d'une liste, ne laissant qu'une seule copie de chaque élément. Les éléments peuvent être des nombres ou des chaînes de caractères.


```python
def justOne(list):
  new_list=[]
  for element in list:
    if element not in new_list:
      new_list.append(element)
  return new_list
```

```python
justOne([1,8,7,5,1,"hello","John","hello"])
```

# Exercice 5

Écrivez une fonction checkType() qui vérifie que la liste fournie en argument ne contient que des éléments du même type et renvoie True ou False selon le cas.



```python
def checkType(list):
  typeToFit=type(list[0])
  check=True
  for element in list[1:]:
    if type(element)!=typeToFit:
      check=False
  return check
```

```python
checkType([1,7,5,8,9,6,4,8,5,3])
```

```python
checkType([1,7,5,"hello",8,6,5,7])
```

1. Écrivez une fonction appelée phone() qui, lorsqu'elle est exécutée, demande à l'utilisateur de remplir le répertoire téléphonique avec les noms et numéros de personnes qu'il connaît. Il peut en ajouter autant qu'il le souhaite.
2. À la fin, la fonction renvoie le nombre de personnes ajoutées.

```python
def phone():
    phoneBook={}
    x=0
    while x!="stop":
        name=input("Quel est le nom de la personne ?")
        tel=input("Quel est son numéro de téléphone ?")
        phoneBook[name]=tel
        x=input("Voulez vous arrêter ? ")
    return len(phoneBook)
```
