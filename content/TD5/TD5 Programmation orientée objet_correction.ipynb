2. Créez une classe Patient prenant comme paramètres l'identifiant du patient, sa date de naissance, une liste de visites et son sexe. Ajoutez les fonctions :
- add_visit, qui ajoute une visite au patient.
- str of Patient, qui affiche l'identifiant, la date de naissance et le sexe du patient.

```python
class Patient():
    def __init__(self,ID,dateBirth,sex):
        self.id=ID
        self.birth=dateBirth
        self.sex=sex
        self.visites=[]
    def get_ID(self):
        return self.id
    def get_birth(self):
        return self.dateBirth
    def get_sex(self):
        return self.sex
    def ajouter_visite(self,visite):
        if visite not in self.visites and visite.get_patient()==self:
            self.visites.append(visite)
    def __str__(self):
        return "ID : %d\nDate de naissance : %s\nGenre : %s"%(self.id,self.birth,self.sex)
```
