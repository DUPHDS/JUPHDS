You are going to work on the "Iris.csv" file.
1. Use panda to import the file.


```python
import pandas as pd 

chemin ="Iris.csv"

iris=pd.read_csv(chemin)
```

2. For each species, calculate the number of individuals.


```python
species={}

for i in iris["Species"]:
    if i not in species:
        species[i]=0
    species[i]+=1

print(species)
```

3. For each species, calculate the average of each parameter: SepalLengthCm, SepalWidthCm, PetalLengthCm, PetalWidthCm.


```python
average={}

for index in range(len(iris["Id"])):
    if iris["Species"][index] not in average:
        average[iris["Species"][index]]={"SepalLengthCm":0, "SepalWidthCm":0, "PetalLengthCm":0, "PetalWidthCm":0}
    for element in average[iris["Species"][index]]:
        average[iris["Species"][index]][element]+=iris[element][index]/species[iris["Species"][index]]

print(average)
```

# Exercice 2

Write a function that displays the multiplication table for a number entered by the user. The
function should generate an error if the number is not between 0 and 10. The function must
also handle the possibility that the user may have entered a string of characters and be
asked to start again.

```python
def multiplicationTable():
    rep=False
    while rep==False:
        try:
            number=int(input("Which mutiplication table do you want ? "))
        except ValueError:
            print("This value is not a number.")
        else:
            rep=True
    if number<0 or number>10:
        raise Exception("The value must be between O and 10.")
    for i in range(0,11):
        print(i*number)
```

1. Write an item class, for which you can specify a name, item type, price and quantity.


```python
class Item():
    def __init__(self,name,itemType,price,quantity):
        self.name=name
        self.itemType=itemType
        self.price=price
        self.quantity=quantity
    def get_name(self):
        return self.name
    def get_itemType(self):
        return self.itemType
    def get_price(self):
        return self.price
    def get_quantity(self):
        return self.quantity
    def __str__(self):
        return self.name
        
```

2. Write a store class with a name and a list of items for sale. This class must be able to add an item to its sales list, sell and replenish an item already in its list. The printout of this class shows the types of items on sale in the store and the quantity remaining. Note that if an item is sold out, it can no longer be sold.

```python
class Store():
    def __init__(self,name):
        self.name=name
        self.itemsSold=[]
    def get_name(self):
        return self.name
    def get_item_sold(self):
        return self.itemsSold
    def addItem(self,item):
        if item not in self.itemsSold:
            self.itemsSold.append(item)
    def sellItem(self,item):
        if item.get_quantity()>0 and item in self.itemsSold:
            item.quantity-=1
        else:
            print("This item is not for sale.")
    def replenishItem(self,item,quantity):
        if item in self.itemsSold:
            item.quantity+= quantity
    def __str__(self):
        return("Name %s \nItems sold : %s"%(self.name,str([item.get_name() for item in self.itemsSold])))
    def __len__(self):
        return len(self.itemsSold)
```

```python
item1=Item("Tomatoes","Fruit",2,2)
```

```python
myStore=Store("Marina's Store")
myStore.addItem(item1)
print(myStore)
```

```python
myStore.sellItem(item1)
item1.get_quantity()
```

```python
myStore.replenishItem(item1,10)
```
