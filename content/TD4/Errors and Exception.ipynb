# Errors

In Python, it's sometimes useful to handle errors to prevent the program from crashing.

```python
path="Test_file.txt"

file=open(path,"w")
file.write("What I want to write.")
file.close()
```

```python
file=open(path,"r")
file.write("No, this is what I want")
```

# Try and Except

To intercept this type of error, we use try and except. The first “Try” block will serve as a test, with the lines of code that may generate the error to be intercepted indicated here. Then “Except” followed by the name of the error that may be raised follows this block. Inside, we indicate that if this error is detected, we want to display something, but not that the program stops as a result of this error, but rather who handles it.

```python
try:
    file=open(path,"r")
    file.write("No, this is what I want")
except IOError:
    print("I intercept this exception !")
```

```python
try:
    answer=int(input("Choose a number : "))
except ValueError:
    print("This is not a number")
else:
    print("Your number is %d"%(answer))
```

# Raise

Raise allows you to define what is an error in addition to the basic system. Let's say you want to raise an error if the user enters a response that isn't appropriate, you could use raise followed by Exception with the error message to be displayed in brackets. These errors can also be intercepted if required.

```python
def getNumberBetween(num_min,num_max):
    try:
        answer=int(input("Choose a number between %d and %d : "%(num_min,num_max)))
        if answer<num_min or answer>num_max:
            raise Exception("It'not between %d and %d"%(num_min,num_max))
    except ValueError:
        print("This is not a number.")
    else:
        print("Success !")
```

```python
getNumberBetween(0,10)
```

```python
getNumberBetween(0,10)
```

```python
getNumberBetween(0,10)
```
