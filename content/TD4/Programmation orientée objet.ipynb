# Définir une classe

Une classe est une structure de données, un objet. Les listes, les entiers et les chaînes de caractères sont tous des classes, des structures.

```python
students=["Max","Monica"]
students.append("Eric")

print(students)
```

Vous pouvez créer vos propres classes pour définir des éléments plus spécifiques. Par exemple, ici, nous voulons créer une classe « étudiant ». Pour ce faire, nous utilisons « class » suivi du nom de la classe, qui commence toujours par une majuscule.

Ensuite, nous définissons la fonction « init », c'est-à-dire la classe qui indiquera comment la classe sera créée et quelles informations elle aura besoin. Ici, pour créer un étudiant, nous avons besoin du prénom et du nom de l'étudiant. Le premier paramètre (qui doit toujours être présent) est self, pour définir que nous voulons créer l'élément par lui-même. Dans cette init, nous allons définir les variables spécifiques à notre objet, ici il aura son propre prénom et son propre nom, définis avec « self. » suivi du nom de la variable.

Après cette fonction, il est possible d'ajouter un certain nombre d'autres fonctions, telles que get_name pour afficher le prénom et le nom de la personne. Notez que seul self est utilisé comme paramètre, ce qui signifie que la fonction n'a besoin de rien d'autre qu'elle-même pour fonctionner.

```python
class Etudiant():
    def __init__(self,fn,ln):
        self.prenom=fn
        self.nom=ln
    def get_name(self):
        print(self.prenom+" "+self.nom)
```

Maintenant que la classe a été créée, nous pouvons créer des instances de celle-ci. Ici, je vais créer la variable eric, qui est un objet Student, avec son prénom et son nom comme arguments. Notez que cette création correspond à init, mais sans l'argument self, car cet argument est donné lorsque la fonction est appelée.

```python
eric=Etudiant("Eric","Smith")
```

Via cette instance, nommée Eric, je peux appeler les différentes fonctions de la classe à laquelle elle appartient, en l'occurrence la classe Student. Je vais donc appeler get_name. Comme indiqué ci-dessus, cette fonction n'a pas d'argument, car le self ne doit pas être indiqué lors de l'utilisation des fonctions, mais uniquement lors de leur création.

```python
eric.get_name()
```

# Exemple 1

Voici un exemple d'une classe nommée PhoneBook. La classe sert de répertoire et est initialisée avec un dictionnaire vide et aucun autre paramètre, ce qui indique que lorsque nous voulons créer une instance de cette classe, il suffit d'appeler PhoneBook().

Pour ajouter des entrées à ce répertoire, nous utiliserons la fonction add ci-dessous, où nous aurons besoin d'un nom et d'un numéro de téléphone comme arguments.

La fonction get sert ici à récupérer le nom d'une personne dans le répertoire téléphonique.

Enfin, « \_str\_ » est la fonction qui permet à votre classe d'afficher quelque chose lorsque vous utilisez print. Sinon, print(object) ne renverra que son adresse. « \_len\_ » est la fonction qui définit ce qui est renvoyé par l'ordinateur lorsque len(object) est utilisé. Cette fonction n'est pas nécessairement obligatoire, mais peut être utile. Ici, par exemple, la longueur d'un PhoneBook renverra le nombre de personnes entrées dans le répertoire.

```python
class PhoneBook():
    def __init__(self):
        self.entries={}
    def add(self,name,phoneNumber):
        self.entries[name]=phoneNumber
    def get(self,name):
        if name in self.entries:
            return self.entries[name]
        else:
            return None
    def __str__(self):
        return "PhoneBook : " + str(self.entries)
    def __len__(self):
        return len(self.entries)
```

Pour tester tous ces paramètres, nous allons créer un répertoire appelé book, qui est un objet de type PhoneBook.

Nous utilisons ensuite la fonction add pour ajouter Doe et Miller.

Nous pouvons alors voir combien de personnes ont été ajoutées et à quoi ressemble notre répertoire.

```python
book= PhoneBook()
book.add("Doe","85687890")
book.add("Miller","7655432")

print(len(book))
print(book)
```
