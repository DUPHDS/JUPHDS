# Erreurs

En Python, il est parfois utile de gérer les erreurs afin d'éviter que le programme ne plante.

```python
path="Test_file.txt"

file=open(path,"w")
file.write("Ce que je veux écrire.")
file.close()
```

```python
file=open(path,"r")
file.write("Non, c'est ce que je veux écrire")
```

# Try et Except

Pour intercepter ce type d'erreur, nous utilisons try et except. Le premier bloc « Try » servira de test, avec les lignes de code susceptibles de générer l'erreur à intercepter indiquées ici. Ensuite, « Except » suivi du nom de l'erreur susceptible d'être levée suit ce bloc. À l'intérieur, nous indiquons que si cette erreur est détectée, nous voulons afficher quelque chose, mais pas que le programme s'arrête à cause de cette erreur, mais plutôt qui la gère.

```python
try:
    file=open(path,"r")
    file.write("Non, c'est ce que je veux écrire")
except IOError:
    print("J'intercepte cette exception !")
```

```python
try:
    answer=int(input("Choisis un nombre : "))
except ValueError:
    print("Ce n'est pas un nombre")
else:
    print("Ton nombre est %d"%(answer))
```

# Raise

Raise vous permet de définir ce qui constitue une erreur en plus du système de base. Supposons que vous souhaitiez générer une erreur si l'utilisateur saisit une réponse inappropriée. Vous pouvez utiliser raise suivi de Exception avec le message d'erreur à afficher entre parenthèses. Ces erreurs peuvent également être interceptées si nécessaire.

```python
def nombreEntre(num_min,num_max):
    try:
        answer=int(input("Choisir un nombre entre %d et %d : "%(num_min,num_max)))
        if answer<num_min or answer>num_max:
            raise Exception("Ce c'est pas entre %d et %d"%(num_min,num_max))
    except ValueError:
        print("Ce n'est pas un nombre.")
    else:
        print("Bravo !")
```

```python
nombreEntre(0,10)
```

```python
nombreEntre(0,10)
```

```python
nombreEntre(0,10)
```
