# Define a class

A class is a data structure, an object. Lists, integers and character strings are all classes, structures.

```python
students=["Max","Monica"]
students.append("Eric")

print(students)
```

You can create your own classes to define more specific things. For example, here we want to create a student class. To do this, we use “class” followed by the name of the class, always beginning with a capital letter.

Then we define the “init” function, i.e. the class that will indicate how the class will be created and what information it will need. Here, to create a student, we need the student's first and last name. The first parameter (which must always be present) is self, to define that we want to create the element by itself. In this init, we'll define the variables specific to our object, here it will have its own first name and its own name, defined with “self.” followed by the variable name.

After this function, it's possible to add a number of other functions, such as get_name to display the person's first and last names. Note that only self is used as a parameter, meaning that the function needs nothing to function except itself.  

```python
class Student():
    def __init__(self,fn,ln):
        self.firstname=fn
        self.lastname=ln
    def get_name(self):
        print(self.firstname+" "+self.lastname)
```

Now, once the class has been created, we can create instances of it. Here, I'm going to create the variable eric, which is a Student object, with his first name and last name as arguments. Note that this creation corresponds to init, but without the self argument, as this argument is given when the function is called.

```python
eric=Student("Eric","Smith")
```

Via this instance, named Eric, I can call the various functions of the class to which it belongs, in this case the Student class. So I'm going to call get_name, this function, as indicated above, has no argument, as the self is not to be indicated when using functions, only when creating them.

```python
eric.get_name()
```

# Example 1

Here's an example of a class named PhoneBook. The class serves as a directory, and is initialized with an empty dictionary and no other parameters, indicating that when we want to create an instance of this class, we just need to call PhoneBook().

To add entries to this directory, we'll use the add function below, where we'll need a name and phone number as arguments.

The get function here is used to fetch a person's name from the phone book.

Finally, “\_str\_” is the function that allows your class to display something when you use print. Otherwise print(object) will only return its address. “\_len\_” is the function that defines what is returned by the computer when len(object) is used. This function is not necessarily mandatory, but can be useful. Here, for example, the len of a PhoneBook will return the number of people entered in the directory.  

```python
class PhoneBook():
    def __init__(self):
        self.entries={}
    def add(self,name,phoneNumber):
        self.entries[name]=phoneNumber
    def get(self,name):
        if name in self.entries:
            return self.entries[name]
        else:
            return None
    def __str__(self):
        return "PhoneBook : " + str(self.entries)
    def __len__(self):
        return len(self.entries)
```

To test all these parameters, we'll create a directory called book, which is a PhoneBook type object.

We then use the add function to add Doe and Miller,

We can then see how many people have been added and what our directory looks like.

```python
book= PhoneBook()
book.add("Doe","85687890")
book.add("Miller","7655432")

print(len(book))
print(book)
```
