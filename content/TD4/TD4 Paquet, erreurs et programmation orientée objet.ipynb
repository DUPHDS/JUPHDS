# Exercice 1

Vous allez travailler sur le fichier « Iris.csv ».
1. Utilisez panda pour importer le fichier.


```python
import pandas as pd

chemin ="Iris.csv"

iris=pd.read_csv(chemin)
```

2. Pour chaque espèce, calculez le nombre d'individus.


```python
species={}

for i in iris["Species"]:
    if i not in species:
        species[i]=0
    species[i]+=1

print(species)
```

3. Pour chaque espèce, calculez la moyenne de chaque paramètre : SepalLengthCm, SepalWidthCm, PetalLengthCm, PetalWidthCm.


```python
average={}

for index in range(len(iris["Id"])):
    if iris["Species"][index] not in average:
        average[iris["Species"][index]]={"SepalLengthCm":0, "SepalWidthCm":0, "PetalLengthCm":0, "PetalWidthCm":0}
    for element in average[iris["Species"][index]]:
        average[iris["Species"][index]][element]+=iris[element][index]/species[iris["Species"][index]]

print(average)
```

# Exercice 2

Écrivez une fonction qui affiche la table de multiplication d'un nombre saisi par l'utilisateur. La
fonction doit générer une erreur si le nombre n'est pas compris entre 0 et 10. La fonction doit
également gérer le cas où l'utilisateur aurait saisi une chaîne de caractères et lui
demander de recommencer.

```python
def multiplicationTable():
    rep=False
    while rep==False:
        try:
            number=int(input("Quel table de multiplication voulez vous ? "))
        except ValueError:
            print("Cette valeur n'est pas un nombre.")
        else:
            rep=True
    if number<0 or number>10:
        raise Exception("La valeur doit être entre 0 et 10.")
    for i in range(0,11):
        print(i*number)
```

# Exercice 3

1. Écrivez une classe d'articles, pour laquelle vous pouvez spécifier un nom, un type d'article, un prix et une quantité.


```python
class Item():
    def __init__(self,name,itemType,prix,quantite):
        self.name=name
        self.itemType=itemType
        self.prix=prix
        self.quantite=quantite
    def get_name(self):
        return self.name
    def get_itemType(self):
        return self.itemType
    def get_prix(self):
        return self.prix
    def get_quantite(self):
        return self.quantite
    def __str__(self):
        return self.name
```

2. Écrivez une classe « magasin » avec un nom et une liste d'articles à vendre. Cette classe doit pouvoir ajouter un article à sa liste de vente, vendre et réapprovisionner un article déjà présent dans sa liste. L'impression de cette classe affiche les types d'articles en vente dans le magasin et la quantité restante. Notez que si un article est épuisé, il ne peut plus être vendu.

```python
class Magasin():
    def __init__(self,name):
        self.name=name
        self.itemsVendu=[]
    def get_name(self):
        return self.name
    def get_item_vendu(self):
        return self.itemsVendu
    def ajoutItem(self,item):
        if item not in self.itemsVendu:
            self.itemsVendu.append(item)
    def venteItem(self,item):
        if item.get_quantite()>0 and item in self.itemsVendu:
            item.quantite-=1
        else:
            print("Cet item n'est pas à vendre.")
    def rajoutItem(self,item,quantite):
        if item in self.itemsVendu:
            item.quantite+= quantite
    def __str__(self):
        return("Name %s \nItems vendu : %s"%(self.name,str([item.get_name() for item in self.itemsVendu])))
    def __len__(self):
        return len(self.itemsVendu)
```

```python
item1=Item("Tomate","Fruit",2,2)
```

```python
monMagasin=Magasin("Marina's Store")
monMagasin.ajoutItem(item1)
print(monMagasin)
```

```python
monMagasin.venteItem(item1)
item1.get_quantite()
```

```python
monMagasin.rajoutItem(item1,10)
```
