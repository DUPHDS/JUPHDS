# Fichiers

## 1 - Ouvrir un fichier : mode lecture

Pour lire le contenu d'un fichier, nous utiliserons le mode lecture de la fonction open.

La fonction readlines nous permet de lire un fichier ligne par ligne, à l'aide d'une boucle.

N'oubliez pas de fermer le fichier avant d'en lire un nouveau.

La fonction read vous permet de lire l'intégralité du fichier en une seule fois.

```python
path="hello.txt"
```

```python
file=open(path,"r")
```

```python
for line in file.readlines():
    print(line)
```

```python
file.close()
```

```python
with open(path, "w") as f:
    f.write("Bonjour tout le monde !\n")
    f.write("J'ajoute une petite phrase.\n")

file=open(path,"r")
```

```python
file=open(path,"r")
chain=file.read()
print(chain)
```

## 2 - Ouvrir un fichier : mode écriture

Pour écrire un fichier, utilisez le mode écriture. Il suffit d'utiliser la fonction d'écriture pour écrire une chaîne de caractères dans le fichier. Vous pouvez également utiliser with, qui ouvrira le fichier avec un nom spécifique, et dans which vous pouvez spécifier les instructions du programme.

Avertissement : ce mode supprime toutes les informations contenues dans le fichier.

```python
path="new_file.txt"
```

```python
file=open(path,"w") # Le mode « w » supprime tout le contenu du fichier s'il a déjà été créé.
```

```python
file.write("Bonjour tout le monde !")
```

```python
file.close()
```

```python
with open(path,"r")as file:
    print(file.read())
```

## 3 - Ouvrir un fichier : mode ajout

Contrairement au mode écriture, le mode ajout ne supprime pas toutes les informations contenues dans le fichier ouvert. L'objectif est d'ajouter des informations à un fichier qui contient déjà des informations.

```python
path="new_file.txt"
file=open(path,"a")
file.write("\nJ'ajoute juste une petite phrase..")
file.close()
```

```python
with open(path,"r")as file:
    print(file.read())
```

## 4 - Propriétés du fichier

Il existe plusieurs propriétés pour les fichiers : vous pouvez savoir dans quel mode le fichier est ouvert avec mode, savoir si le fichier est fermé avec closed, et également obtenir son nom avec name.

```python
path="new_file.txt"
file=open(path,"r")
```

```python
file.mode
```
