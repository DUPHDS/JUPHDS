# Files

## 1 - Open a file : read mode

To read the contents of a file, we'll use the read mode of the open function.

The readlines function lets us read a file line by line, using a loop.

Don't forget to close the file before reading a new one.

The read function allows you to read the entire file at once.

```python
path="hello.txt"
```

```python
file=open(path,"r")
```

```python
for line in file.readlines():
    print(line)
```

```python
file.close()
```

```python
file=open(path,"r")
chain=file.read()
print(chain)
```

## 2 - Open a file : write mode

To write a file, use the write mode. Simply use the write function to write a character string inside the file. You can also use with as, which will open the file with a specific name, and within with you can specify the program instructions.

Warning: this mode deletes all the information contained in the file.

```python
path="new_file.txt"
```

```python
file=open(path,"w") # "w" mode remove everything from the file if it was already created
```

```python
file.write("Hello everyone !")
```

```python
file.close()
```

```python
with open(path,"r")as file:
    print(file.read())
```

## 3 - Open a file : add mode

Unlike write mode, add mode does not delete all the information contained in the file being opened. The aim is to add information to a file already containing information.

```python
path="new_file.txt"
file=open(path,"a")
file.write("\nI'm adding juste a little sentence.")
file.close()
```

```python
with open(path,"r")as file:
    print(file.read())
```
