# Exercice 2

The "heart.csv" file presents the cardiac information of several kaggle patients, and is available at this
address: https://www.kaggle.com/ronitf/heart-disease-uci . The aim of this exercise is to synthesize
the data.

1. Scan the document and count the number of men and women ("0" for woman and "1" for man in the document).

```python
heartfile=open("heart.csv","r")

sex={"Female":0,"Male":0}

firstLine=True

for line in heartfile.readlines():
    sliceLine=line.split(",")
    if firstLine!=True:
        if sliceLine[1]=="0":
            sex["Female"]+=1
        else:
            sex["Male"]+=1
    firstLine=False
    
print(sex)

heartfile.close()
```

```python
for line in heartfile.readlines():
    sliceLine=line.split(",")
    if sliceLine[1]=="0":
        sex["Female"]+=1
    elif sliceLine[1]=="1":
        sex["Male"]+=1
```

2. The third column, "cp", indicates the type of chest pain experienced by the patient. For each
gender, calculate the percentage of each type of pain.

```python
heartfile=open("heart.csv","r")

chest_pain_gender={"Female":{},"Male":{}}

for line in heartfile.readlines():
    sliceLine=line.split(",")
    if sliceLine[1]=="0":
        if sliceLine[2] not in chest_pain_gender["Female"]:
            chest_pain_gender["Female"][sliceLine[2]]=0
        chest_pain_gender["Female"][sliceLine[2]]+=1
    elif sliceLine[1]=="1":
        if sliceLine[2] not in chest_pain_gender["Male"]:
            chest_pain_gender["Male"][sliceLine[2]]=0
        chest_pain_gender["Male"][sliceLine[2]]+=1

print(chest_pain_gender)
file.close()
```

```python
heartfile=open("heart.csv","r")

chest_pain_gender={}

for line in heartfile.readlines():
    sliceLine=line.split(",")
    if sliceLine[1] not in chest_pain_gender and sliceLine[1]!="sex":
        chest_pain_gender[sliceLine[1]]={}
    elif sliceLine[1]=="sex":
        continue
    if sliceLine[2] not in chest_pain_gender[sliceLine[1]]:
        chest_pain_gender[sliceLine[1]][sliceLine[2]]=0
    chest_pain_gender[sliceLine[1]][sliceLine[2]]+=1

print(chest_pain_gender)
file.close()
```

```python
# I have sex and i have chest_pain_gender

for s in chest_pain_gender:
    if s=="0":
        print("FEMALE :\n")
        for item in chest_pain_gender[s]:
            percentage=(chest_pain_gender[s][item]/sex["Female"])*100
            print("Chest pain type : %s Percentage %f"%(item,percentage))
    else:
        print("MALE :\n")
        for item in chest_pain_gender[s]:
            percentage=(chest_pain_gender[s][item]/sex["Male"])*100
            print("Chest pain type : %s Percentage %f"%(item,percentage))
```
