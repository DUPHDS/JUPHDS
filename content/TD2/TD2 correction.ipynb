# Exercice 2

Le fichier « heart.csv » présente les informations cardiaques de plusieurs patients Kaggle et est disponible à l'adresse suivante :
https://www.kaggle.com/ronitf/heart-disease-uci . L'objectif de cet exercice est de synthétiser
les données.

1. Parcourez le document et comptez le nombre d'hommes et de femmes (« 0 » pour les femmes et « 1 » pour les hommes dans le document).

```python
heartfile=open("heart.csv","r")

sex={"Femme":0,"Homme":0}

firstLine=True

for line in heartfile.readlines():
    sliceLine=line.split(",")
    if firstLine!=True:
        if sliceLine[1]=="0":
            sex["Femme"]+=1
        else:
            sex["Homme"]+=1
    firstLine=False

print(sex)

heartfile.close()
```

```python
for line in heartfile.readlines():
    sliceLine=line.split(",")
    if sliceLine[1]=="0":
        sex["Femme"]+=1
    elif sliceLine[1]=="1":
        sex["Homme"]+=1
```

2. La troisième colonne, « cp », indique le type de douleur thoracique ressentie par le patient. Pour chaque
sexe, calculez le pourcentage de chaque type de douleur.

```python
heartfile=open("heart.csv","r")

chest_pain_gender={"Femme":{},"Male":{}}

for line in heartfile.readlines():
    sliceLine=line.split(",")
    if sliceLine[1]=="0":
        if sliceLine[2] not in chest_pain_gender["Femme"]:
            chest_pain_gender["Femme"][sliceLine[2]]=0
        chest_pain_gender["Femme"][sliceLine[2]]+=1
    elif sliceLine[1]=="1":
        if sliceLine[2] not in chest_pain_gender["Homme"]:
            chest_pain_gender["Homme"][sliceLine[2]]=0
        chest_pain_gender["Homme"][sliceLine[2]]+=1

print(chest_pain_gender)
file.close()
```

```python
heartfile=open("heart.csv","r")

chest_pain_gender={}

for line in heartfile.readlines():
    sliceLine=line.split(",")
    if sliceLine[1] not in chest_pain_gender and sliceLine[1]!="sex":
        chest_pain_gender[sliceLine[1]]={}
    elif sliceLine[1]=="sex":
        continue
    if sliceLine[2] not in chest_pain_gender[sliceLine[1]]:
        chest_pain_gender[sliceLine[1]][sliceLine[2]]=0
    chest_pain_gender[sliceLine[1]][sliceLine[2]]+=1

print(chest_pain_gender)
file.close()
```

```python
# J'ai les genres et j'ai les douleurs thoraciques liées au genre.

for s in chest_pain_gender:
    if s=="0":
        print("FEMALE :\n")
        for item in chest_pain_gender[s]:
            percentage=(chest_pain_gender[s][item]/sex["Female"])*100
            print("Chest pain type : %s Percentage %f"%(item,percentage))
    else:
        print("MALE :\n")
        for item in chest_pain_gender[s]:
            percentage=(chest_pain_gender[s][item]/sex["Male"])*100
            print("Chest pain type : %s Percentage %f"%(item,percentage))
```
