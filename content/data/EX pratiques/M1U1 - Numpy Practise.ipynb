#### Comment résoudre un système d'équations linéaires à l'aide de NumPy ?
#### Exemple : résolvez x et y dans le système :
#### 2x+3y=8
#### 5x+7y=19

```python
# Coefficient Matrix
A = np.array([[2,3], [5,7]])
```

```python
#print(A)
```

```python
#Constant Matrix
B=np.array([8,19])
```

```python
#print(B)
```

```python
solution = np.linalg.solve(A,B)
```

```python
#print(solution)
```
