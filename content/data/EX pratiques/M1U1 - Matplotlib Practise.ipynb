```python
import matplotlib.pyplot as plt
```

#### Problème 1 : Nous disposons des données mensuelles relatives au chiffre d'affaires d'une entreprise pour la période allant de janvier à juin :
#### Mois :  ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
#### Chiffre d'affaires (en milliers) : [10, 20, 15, 25, 30, 40]
#### Tâche : créer un graphique linéaire pour les données de ventes avec les étiquettes, le titre et la grille appropriés.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [10, 20, 15, 25, 30, 40]
```

```python
plt.plot(months, sales, marker='o', linestyle='-', color='b', label="Tendance des ventes")
plt.xlabel("Mois")
plt.ylabel("Ventes (en milliers)")
plt.title("Données mensuelles sur les ventes")
plt.legend()
plt.grid(True)
```

```python
plt.show()
```

#### Problème 2 :  Graphique à barres pour les notes des élèves
#### Un enseignant note les résultats des élèves à un test :
#### Élèves : [ ["Alice", "Bob", "Charlie", "David", "Eve"]
#### Notes : [85, 90, 78, 92, 88]
#### Tâche : créer un diagramme à barres pour visualiser les notes des élèves.

```python
students = ["Alice", "Bob", "Charlie", "David", "Eve"]
scores = [85, 90, 78, 92, 88]

plt.bar(students, scores, color=["red", "blue", "green", "purple", "orange"])
plt.xlabel("Etudiants")
plt.ylabel("Notes")
plt.title("Notes des étudiants")
plt.show()
```

#### Problème 3 : Diagramme de dispersion pour la taille et le poids
#### Nous disposons de la taille et du poids de 10 personnes :
#### Taille (cm) : [150, 160, 165, 170, 175, 180, 185, 190, 195, 200]
#### Poids (kg) : [50, 55, 58, 60, 65, 70, 75, 80, 85, 90]
#### Tâche : créer un nuage de points pour visualiser la taille par rapport au poids.

```python
import numpy as np

height = [150, 160, 165, 170, 175, 180, 185, 190, 195, 200]
weight = [50, 55, 58, 60, 65, 70, 75, 80, 85, 90]

plt.scatter(height, weight, color="blue", marker="o")
plt.xlabel("Hauteur (cm)")
plt.ylabel("Poids (kg)")
plt.title("Hauteur vs Poids Scatter Plot")
plt.show()
```

#### Problème 4 : Histogramme de la répartition par âge
#### Une enquête a permis de recueillir l'âge de 100 personnes. Générez un histogramme pour montrer la répartition par âge.
#### Tâche : Créez un histogramme avec 10 classes et des étiquettes appropriées.

```python
ages = np.random.randint(10, 60, 100)  # Générer 100 âges aléatoires compris entre 10 et 60 ans.

plt.hist(ages, bins=10, color="cyan", edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Fréquence")
plt.title("Distribution de l'âge Histogram")
plt.show()
```

#### Problème 5 : Diagramme circulaire représentant les parts de marché
#### Les parts de marché des différentes entreprises sont les suivantes :
#### Entreprises :  ["Apple", "Samsung", "Xiaomi", "Oppo", "Vivo"]
#### Parts de marché (%) : [30, 25, 15, 20, 10]
#### Tâche : créer un graphique circulaire pour visualiser les parts de marché.

```python
companies = ["Apple", "Samsung", "Xiaomi", "Oppo", "Vivo"]
market_share = [30, 25, 15, 20, 10]
colors = ["red", "blue", "green", "purple", "orange"]

plt.pie(market_share, labels=companies, colors=colors, autopct="%1.1f%%", startangle=140)
plt.title("Part de marché des smartphones")
plt.show()
```

#### Problème 6 : graphiques à lignes multiples (comparaison des ventes)
#### Nous disposons des données de vente de deux magasins différents sur six mois.
#### Mois : ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
#### Ventes du magasin A : [12, 18, 25, 30, 35, 40]
#### Ventes du magasin B : [10, 15, 22, 28, 32, 38]
#### Tâche : tracez deux graphiques linéaires sur la même figure afin de comparer les ventes.


```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
store_a_sales = [12, 18, 25, 30, 35, 40]
store_b_sales = [10, 15, 22, 28, 32, 38]

plt.plot(months, store_a_sales, marker='o', linestyle='-', color='blue', label="Ventes du magasin A")
plt.plot(months, store_b_sales, marker='s', linestyle='--', color='red', label="Ventes du magasin B")
plt.xlabel("Mois")
plt.ylabel("Ventes (en milliers)")
plt.title("Comparaison des ventes")
plt.legend()
plt.grid(True)
plt.show()
```
