#### How do you solve a system of linear equations using NumPy?
#### Example: Solve for x and y in the system:
#### 2x+3y=8
#### 5x+7y=19

```python
# Coefficient Matrix
A = np.array([[2,3], [5,7]])
```

```python
#print(A)
```

```python
#Constant Matrix
B=np.array([8,19])
```

```python
#print(B)
```

```python
solution = np.linalg.solve(A,B)
```
