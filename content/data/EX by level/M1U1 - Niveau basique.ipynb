« Chargez un ensemble de données sur la santé publique et calculez les statistiques de base (moyenne, médiane et écart type) pour l'âge. »


```python
import pandas as pd
# Load the dataset
df = pd.read_csv("healthcare-dataset-stroke-data.csv")
# Calculate basic statistics
mean_age = df["age"].mean()
median_age = df["age"].median()
std_age = df["age"].std()

print(f"Age moyen: {mean_age}")
print(f"Age médian: {median_age}")
print(f"Écart type de l'âge: {std_age}")
```
