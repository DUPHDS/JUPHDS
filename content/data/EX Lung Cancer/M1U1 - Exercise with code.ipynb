#1. Build a logistic regression model to predict lung cancer based on the dataset (after encoding categorical variables).

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
```

```python
#encoded_data = df.copy()    # Make copy of original data
#categorical_columns = df.select_dtypes(include=['object']).columns  # check type of columns that have a data type 
                                             #of object, meaning they contain textual/categorical data.
#le = LabelEncoder()  #LabelEncoder converts categorical values into numerical labels.
#for col in categorical_columns:  #Looping Over Categorical Columns and Encoding
    #encoded_data[col] = le.fit_transform(df[col])  #Applies LabelEncoder to transform text values into numbers.
```

```python
# Encode categorical variables
encoded_data = df.copy()
categorical_columns = df.select_dtypes(include=['object']).columns
le = LabelEncoder()
for col in categorical_columns:
    encoded_data[col] = le.fit_transform(df[col])
```

```python
# Define features (independent features) and target(dependent feature)
#X = encoded_data.drop(columns=['LUNG_CANCER'])    #Remove the target variable
#y = encoded_data['LUNG_CANCER']
```

```python
# Define features and target
X = encoded_data.drop(columns=['LUNG_CANCER'])
y = encoded_data['LUNG_CANCER']
```

```python
#Why Do We Split Data?

    #Machine learning models learn from training data (X_train, y_train).
    #Testing data (X_test, y_test) is used to evaluate the model's performance on unseen data.
    #This prevents overfitting (when a model memorizes training data but fails on new data).
  #train_test_split():-This function from sklearn.model_selection randomly splits the dataset into 
    #training and testing sets.  
  #X=Input features (independent variables)
#y=Target variable (dependent variable)
#test_size=0.2=20% of data goes into the test set, 80% into training
#random_state=42   = Ensures reproducibility (same split every time)
    
    
    
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=50)
```

```python
#Train LR model:- trains, and fits a Logistic Regression model using X_train (input features) 
#and y_train (target variable).
#.fit(X_train, y_train) trains the model using training data (X_train) and labels (y_train).
#The model learns patterns in the training data and optimizes weights (coefficients) to make accurate predictions.
```

```python
# Train logistic regression model
model = LogisticRegression()    #initializes a logistic regression model.
model.fit(X_train, y_train)     #trains the model using labeled data.
```

```python
#Accuracy:evaluates the performance of the trained logistic regression model by calculating its accuracy 
#on the test data (X_test, y_test).
```

```python
# Check model accuracy
accuracy = model.score(X_test, y_test)  #.score() method in Scikit-Learn computes accuracy for classification models.
accuracy      #(Correct Predictions) / (Total Predictions).
```

```python
#2.Identify the top 3 features most strongly associated with lung cancer using feature importance.

feature_importance = pd.DataFrame({'Feature': X.columns, 'Importance': abs(model.coef_[0])}).sort_values(by='Importance', ascending=False)
feature_importance.head(3)
```

```python
#chi-square test:- to check if there is a statistical association between smoking and lung cancer. 
#If p < 0.05 → Strong evidence that smoking and lung cancer are related.
#If p ≥ 0.05 → No strong evidence of association
```

```python
# 3. Perform a chi-square test to check the association between 'SMOKING' and 'LUNG_CANCER'.

from scipy.stats import chi2_contingency

table = pd.crosstab(df['SMOKING'], df['LUNG_CANCER'])
chi2, p, dof, expected = chi2_contingency(table)
p
```

```python
# 4. Use a heatmap to visualize correlations between encoded variables in the dataset.

import numpy as np

# Compute the correlation matrix
correlation_matrix = encoded_data.corr()

# Plot the heatmap
plt.figure(figsize=(12, 8))
plt.imshow(correlation_matrix, cmap='coolwarm', interpolation='none')
plt.colorbar(label='Correlation Coefficient')

# Add annotations
for i in range(len(correlation_matrix)):
    for j in range(len(correlation_matrix)):
        plt.text(j, i, f"{correlation_matrix.iloc[i, j]:.2f}", ha='center', va='center', color='black', fontsize=8)

# Set tick labels
plt.xticks(range(len(correlation_matrix)), correlation_matrix.columns, rotation=90)
plt.yticks(range(len(correlation_matrix)), correlation_matrix.columns)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# 5. Build a decision tree classifier to predict lung cancer and evaluate its accuracy.

from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_accuracy = dt_model.score(X_test, y_test)
dt_accuracy
```

```python
#Cross-validation is a technique used to evaluate the generalizability of a machine learning model.
#It works by splitting the dataset into multiple subsets (folds), training the model on some folds and 
#testing it on others.
#Cross-validation helps to reduce overfitting, as the model is trained and evaluated on different data splits.

#cv=5: Specifies that the cross-validation should use 5 folds. This means the data is split into 5 equal subsets. 
#The model will be trained 5 times, each time using 4 subsets for training and 1 subset for testing. 
#The model's performance is then averaged across the 5 folds.

#Better estimation of model performance: Cross-validation provides a more reliable estimate of how well your 
#model will perform on new data, compared to using a single train-test split.
```

```python
# 6. Perform cross-validation to evaluate the logistic regression model.

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(model, X, y, cv=5)
cv_scores.mean()
```
