1. Using a loop, write the multiplication table for 8.

```python
# Create a list from 0 to 10
# Iterate on this list
# Multiply each number by 8
# Print it

a=range(0,11)
for item in a:
    print(item*8)
```

```python
for item in range(0,11):
    print(item*8)
```

```python
for item in [x*8 for x in range(0,11)]:
    print(item)
```

```python
lst_by_8=[x*8 for x in range(1,11)]
print(lst_by_8)
```

2. Create a list from 0 to 40, loop through it and count the number of multiples of 7.

```python
# Create a variable that count multiples of 7
# Create a loop that goes from 0 to 40
# At each iteration if it's a multiple I add 1 to my variable 

numberMultiple7=0

for item in range(0,40):
    if item%7==0:
        numberMultiple7+=1
print(numberMultiple7)
```

3. Using a dictionary and a list of 11 to 188 in steps of 3, separate even and odd numbers. If the
number is a multiple of 10, print it out.

```python
# Create a dictionnary with 2 keys associated with empty list : 
#"Odd" and "Even"
# Create a loop from 11 to 188 with a step of 3
# For each iteration, if it's an odd number -> add it to the 
# dictionnary with the key "Odd"
# For each iteration, if it's an even number -> add it to the 
# dictionnary with the key "Even"
# If the number is a multiple of 10 -> print it

numberDict={"Odd":[],"Even":[]}

for item in range(11,189,3):
    if item%2==0:
        numberDict["Even"].append(item)
    else:
        numberDict["Odd"].append(item)
    if item%10==0:
        print(item)

print(numberDict)
```

# Exercice 2

Write a program that takes sentences from the user as input and only stops when the user has used
the "a" character 10 times.

```python
# Create a while loop that stops only when the user added 10 a as input
# Each iteration get the sentence from the user
# Count the number of "a" in this sentence

numberA=0

while numberA<10:
    sentence=input("What do you want to say ?")
    for letter in sentence:
        if letter=="a":
            numberA+=1
    print(numberA)
```

# Exercice 3

Write a program that fills a directory (dictionary) with telephone numbers. The user must enter a name
and number each time he wants to add a new person. He can also choose to stop adding people and
interrupt the program.

```python
# While loop until the user wants to stop
# Dictionnary that is empty at the beginning
# For every loop, we can ask if the user wants to stop or add 
# a new person
# Ask the answer to the user : if he writes "STOP" the program stop. 
# For every person, we need to ask a name and a telephone number.
# Those informations will be stored in the dictionnary.

phoneBook={}
a=True

while a:
    answerStop=input("Do you want to stop ? [yes/no]")
    if answerStop=="yes":
        a=False
    else:
        name=input("What is the name ?")
        number=input("What is the number ?")
        phoneBook[name]=number
        print("Person added successfully !")

print(phoneBook)
```
