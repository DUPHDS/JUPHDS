2. Créez une liste de 0 à 40, parcourez-la et comptez le nombre de multiples de 7.

```python
# Créer une variable qui compte les multiples de 7.
# Créer une boucle qui va de 0 à 40.
# À chaque itération, s'il s'agit d'un multiple, j'ajoute 1 à ma variable.

numberMultiple7=0

for item in range(0,40):
    if item%7==0:
        numberMultiple7+=1
print(numberMultiple7)
```

3. À l'aide d'un dictionnaire et d'une liste allant de 11 à 188 par incréments de 3, séparez les nombres pairs et impairs. Si le
nombre est un multiple de 10, affichez-le.

```python
# Créer un dictionnaire avec 2 clés associées à une liste vide :
#« Impair » et « Pair »
# Créer une boucle de 11 à 188 avec un pas de 3
# Pour chaque itération, si c'est un nombre impair -> l'ajouter au
# dictionnaire avec la clé « Impair »
# Pour chaque itération, s'il s'agit d'un nombre pair -> l'ajouter au
# dictionnaire avec la clé « Pair »
# Si le nombre est un multiple de 10 -> l'imprimer

numberDict={"Impair":[],"Pair":[]}

for item in range(11,189,3):
    if item%2==0:
        numberDict["Pair"].append(item)
    else:
        numberDict["Impair"].append(item)
    if item%10==0:
        print(item)

print(numberDict)
```

# Exercice 2

Écrivez un programme qui prend en entrée les phrases saisies par l'utilisateur et ne s'arrête que lorsque l'utilisateur a utilisé
le caractère « a » 10 fois.

```python
# Créer une boucle while qui s'arrête uniquement lorsque l'utilisateur a ajouté 10 « a » en entrée.
# Chaque itération récupère la phrase de l'utilisateur.
# Compter le nombre de « a » dans cette phrase.

numberA=0

while numberA<10:
    sentence=input("Que voulez vous dire ?")
    for letter in sentence:
        if letter=="a":
            numberA+=1
    print(numberA)
```

# Exercice 3

Écrivez un programme qui remplit un répertoire (dictionnaire) avec des numéros de téléphone. L'utilisateur doit entrer un nom
et un numéro chaque fois qu'il souhaite ajouter une nouvelle personne. Il peut également choisir d'arrêter d'ajouter des personnes et d'
interrompre le programme.

```python
# Boucle while jusqu'à ce que l'utilisateur souhaite arrêter
# Dictionnaire vide au début
# À chaque boucle, nous pouvons demander à l'utilisateur s'il souhaite arrêter ou ajouter
# une nouvelle personne
# Demander la réponse à l'utilisateur : s'il écrit « STOP », le programme s'arrête.
# Pour chaque personne, nous devons demander un nom et un numéro de téléphone.
# Ces informations seront stockées dans le dictionnaire.

phoneBook={}
a=True

while a:
    answerStop=input("Voulez vous arrêter ? [oui/non]")
    if answerStop=="oui":
        a=False
    else:
        name=input("Quel est le nom de a personne à ajouter ?")
        number=input("Le numéro de téléphone ?")
        phoneBook[name]=number
        print("La personne a bien été ajoutée !")

print(phoneBook)
```
