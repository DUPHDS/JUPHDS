## 3.5 - Formatage d'impression

Les variables peuvent être insérées dans des chaînes de plusieurs façons. Supposons que vous ayez une chaîne que vous souhaitez compléter avec une variable qui sera différente selon le contexte, par exemple le nom d'une personne. Il est alors possible d'indiquer un emplacement dans la chaîne où vous souhaitez que la variable soit insérée. La première méthode est le format. Utilisez des accolades pour indiquer l'emplacement où la variable doit être insérée, puis ajoutez .format() avec la variable entre parenthèses. Si vous souhaitez ajouter plusieurs variables, indiquez-les simplement dans l'ordre dans lequel elles apparaissent dans la chaîne.

```python
"Bonjour, intégre ton prénom {}".format("Marina")
```

```python
"Bonjour, intégre tes 3 plats favorits {1} {2} {0}".format("Pizza","Pâtes","Salade")
```

La deuxième méthode consiste à utiliser %s, où vous indiquez le type de variable à insérer dans la chaîne : %s pour les chaînes, %d pour les entiers et %f pour les flottants. Avec %f, vous pouvez spécifier 2 paramètres supplémentaires, le nombre de caractères à insérer avec le chiffre avec %10f, qui indique que le chiffre sera composé de 10 caractères (il ajoutera des 0 pour compléter si le chiffre ne comporte pas 10 caractères), et également spécifier le nombre de décimales avec %.2f, qui indique que le chiffre aura deux chiffres après la virgule décimale.

```python
myName="Marina"
```

```python
print("Bonjour, intégre ton prénom : %s"%(myName))
```

```python
a="maïs"
b="salade"
```

```python
print("Des ingrédients : %s et %s"%(a,b))
```

```python
print("Vous pouvez afficher des nombres aussi : %d"%(3))
```

```python
print("Mais pour afficher des chiffres à virgule n'utilisez pas d : %d mais f : %f"%(3.42,3.76))
```

```python
print("Vous pouvez préciser le nombre de chiffre après la virgule : %.2f"%(3.765))
```

```python
print("Vous pouvez préciser le nombre minimum de caractères : %10f"%(3))
```

```python
print("Vous pouvez spécifier le nombre de caractères et le nombre de décimales: %10.2f"%(3))
```

Vous pouvez également insérer des tabulations dans la chaîne avec \t et des sauts de ligne avec \n.

# 8 -  Fichiers

Il existe plusieurs façons de créer des documents en Python. Sous Jupyter Notebook, la première consiste à utiliser %%writefile, auquel vous associez le nom du fichier avec son extension, dans ce cas test.txt. Une fois la cellule exécutée, le fichier sera créé au même endroit que votre fichier Jupyter Notebook.



```python
%%writefile test.txt
Bonjour, c'est un fichier test rapide.
2ème ligne.
```

Le premier élément de cette méthode est le chemin d'accès au fichier, puis le second indique comment le fichier est ouvert : « w » pour le mode écriture indique que vous souhaitez supprimer le contenu du fichier et y écrire de nouvelles informations, « a » pour le mode ajout indique que vous souhaitez ajouter du contenu au fichier, et « r » pour le mode lecture indique que vous souhaitez simplement lire les données qu'il contient.

```python
myfile = open("Text/test.txt","w")
```

La commande pwd vous indique le chemin d'accès au dossier dans lequel vous vous trouvez, ce qui est utile à savoir si vous êtes perdu.

```python
pwd
```

## 8.1 - Ouverture d'un fichier

La méthode open permet d'ouvrir un fichier ; le mode par défaut est « r ».
La méthode read lit l'intégralité du fichier et affiche son contenu.

```python
my_file = open('test.txt')
```

```python
my_file.read()
```

Cependant, si vous l'exécutez deux fois de suite, le fichier considérera que vous êtes à la fin du document et n'affichera rien lors de la deuxième exécution. Vous devrez réinitialiser la position du curseur à zéro à l'aide de la méthode seek, le 0 indiquant que vous souhaitez revenir au début du document.

```python
my_file.read()
```

```python
my_file.seek(0)
```

```python
my_file.read()
```

La méthode readlines lit le document ligne par ligne et stocke chaque ligne dans une liste.

```python
my_file.seek(0)
my_file.readlines()
```

Fermez toujours le document pour indiquer que vous avez terminé de le manipuler à l'aide de la méthode close.

```python
my_file.close()
```

## 8.4 - Itération

En utilisant une boucle, vous pouvez parcourir le document. Ici, par exemple, nous parcourons le document test.txt ligne par ligne, chaque ligne étant stockée dans x au fur et à mesure que la boucle progresse. Au premier tour, x est égal à « Première ligne » et au deuxième tour à « Deuxième ligne ». Il n'y a que deux tours car il n'y a que deux lignes.

```python
%%writefile test.txt
Première ligne
Seconde ligne
```

```python
for x in open('test.txt'):
    print(x)
```

# Entrée (input)

La fonction d'entrée (input) est utilisée pour la communication entre l'ordinateur et l'utilisateur. Elle permet de demander à l'utilisateur d'entrer une valeur qui peut ensuite être réutilisée dans le code.

```python
name = input("Dis moi ton nom : ")
```

```python
print(name)
```

Veuillez noter que les valeurs que vous récupérez avec input sont automatiquement des chaînes de caractères. Pour les utiliser comme des nombres, vous devez les transformer en entiers à l'aide de la fonction int.

```python
a= input("Donne moi un chiffre : ")
for i in range(a):
  print(i)
```

```python
a= int(input("Donne moi un chiffre : "))
for i in range(a):
  print(i)
```
