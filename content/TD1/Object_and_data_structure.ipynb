## 3.2 - String basics

There are functions or methods that are specific to certain data types. Here, for example, for the string type, there's the len() method for obtaining the length of the character string.

It is possible to access locations in the character string, for example the character at place 0 of the list a using square brackets: a[0] .

```python
a[0]
```

The place corresponding to -1 is the last character in the string.

```python
a[-1]
```

To obtain the character string from a certain position, we use a colon after the position we're interested in. a[4:] means we want the part of the character string starting at position 4 to the end.

```python
a[4:]
```

To obtain the character string before a certain position, we use a colon before the position of interest. a[:2] means that we want the part of the character string from the beginning of the string to position 2.

```python
a[:2]
```

It's also possible to take only part of a character string by making jumps. For example, if I only want the letter every 2 positions, I'll use a[::2].

```python
a[::2]
```

Note, however, that you can't reassign a position to another letter; this manipulation is possible in list types, but not in strings.

```python
a[0]="x"
```

## 3.5 - Print formatting

Variables can be inserted into strings in a number of ways. Let's say you have a string that you want to complete with a variable that will be different depending on the context, for example the name of a person, then it's possible to indicate a location in the string where you want the variable to be inserted. The first method is format. Use curly brackets to indicate where the variable is to be inserted, and then add .format() with the variable in brackets. If you wish to add more than one variable, simply indicate them in the order in which they appear in the string.

```python
"Hello, insert your name {}".format("Marina")
```

```python
"Hello, insert your 3 favorites food {1} {2} {0}".format("Pizza","Pasta","Salad")
```

The second method is to use %s, where you indicate the type of variable to be inserted in the string: %s for strings, %d for integers, and %f for floats. With %f you can specify 2 additional parameters, the number of characters to be inserted with the digit with %10f, which indicates that the digit will be composed of 10 characters (it will add 0 to complete if the digit is not 10 characters long), and also specify the number of decimal places with %.2f, which indicates that the digit will have two digits after the decimal point.

```python
myName="Marina"
```

```python
print("Hello, insert your name : %s"%(myName))
```

```python
a="corn"
b="lettuce"
```

```python
print("Some ingredients : %s and %s"%(a,b))
```

```python
print("You can print numbers too : %d"%(3))
```

```python
print("But to print floats don't use d : %d but f : %f"%(3.42,3.76))
```

```python
print("You can specify the number of decimals : %.2f"%(3.765))
```

```python
print("You can specify a minimum of number of character : %10f"%(3))
```

```python
print("You can specify a minimum of number of character and the number of decimals: %10.2f"%(3))
```

You can also insert tabs in the string with \t and line breaks with \n.

```python
print("You can include tab in your sentences : \thello\teveryone\t!")
```

The last method uses the f: place the f in front of the string and indicate the variable to be inserted in the string between curly brackets.

```python
name="Marina"
```

```python
print(f"My name is {name}")
```

# 8 - Files

There are several ways of creating documents in Python. Under jupyter notebook, the first is to use %%writefile, to which you associate the name of the file with its extension, in this case test.txt. Once the cell has been executed, the file will be created in the same place as your jupyter notebook file.



```python
%%writefile test.txt
Hello, this is a quick test file.
2nd line
```

The first element of this method is the file path, then the second indicates how the file is opened: “w” for write mode indicates that you want to delete the file contents and write new information inside, “a” for append mode indicates that you want to add contents to the file, and “r” for read mode indicates that you just want to read the data inside.

```python
myfile = open("Text/test.txt","w")
```

The pwd command gives you the path to the folder you're in, which is useful to know if you're lost.

```python
pwd
```

## 8.1 - Opening a file

The open method is used to open a file; the default mode is “r”.
The read method reads the entire file and displays its contents.

```python
my_file = open('test.txt')
```

```python
my_file.read()
```

However, if you run it twice in succession, the file will consider you to be at the end of the document and won't display anything for the second run. You'll need to reset the cursor location to zero using the seek method, the 0 indicating that you want to return to the beginning of the document.

```python
my_file.read()
```

```python
my_file.seek(0)
```

```python
my_file.read()
```

The readlines method reads the document line by line and stores each line in a list.

```python
my_file.seek(0)
my_file.readlines()
```

Always close the document to indicate that you have finished manipulating it with the close method.

```python
my_file.close()
```

## 8.2 - Writing a file

To write to a file, we use the write mode. The write method is used to write a character string to the document. Always specify a string or use str(variable) to transform it. If you want to read the document afterwards, don't forget to close the document and reopen it in read mode.

```python
my_file = open('test.txt','w')
```

```python
my_file.write('This is a new line')
```

```python
my_file.close()
my_file=open("test.txt")
```

```python
my_file.seek(0)
my_file.read()
```

```python
my_file.close()
```

## 8.3 - Adding new lines

You can add several lines in succession.

```python
my_file = open('test.txt','a')
my_file.write('\nThis is text being appended to test.txt')
my_file.write('\nAnd another line here.')
my_file.close()
my_file=open("test.txt")
```

```python
my_file.seek(0)
print(my_file.read())
```

```python
my_file.close()
```

## 8.4 - Iterating

By using a loop, you can browse through the document. Here, for example, we run through the document test.txt line by line, each line being stored in x as the loop progresses. In the first round, x is equal to “First line” and in the second round to “Second line”. There are only two rounds because there are only two lines.

```python
%%writefile test.txt
First Line
Second Line
```

```python
for x in open('test.txt'):
    print(x)
```
